```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
from typing import List, Dict, Any
import ast


# ============================================================
# INVENTRA AI — noyau V0.1
# ============================================================

@dataclass
class ProblemSpec:
    description: str
    domain: str
    objective: str
    users: str
    constraints: List[str]
    success_criteria: List[str]


@dataclass
class Idea:
    name: str
    approach: str
    advantages: List[str]
    limitations: List[str]
    complexity: int  # 1 = faible, 5 = élevée
    template_key: str


@dataclass
class Evaluation:
    idea_name: str
    usefulness: int
    feasibility: int
    testability: int
    simplicity: int
    total_score: float
    reasons: List[str]


@dataclass
class ArchitecturePlan:
    title: str
    components: List[Dict[str, str]]
    data_flow: List[str]
    risks: List[str]
    tests: List[str]


class ProblemAnalyzer:
    """Transforme une demande libre en spécification simple et lisible."""

    def analyze(self, description: str) -> ProblemSpec:
        text = (description or "").strip()
        if len(text) < 12:
            raise ValueError("Décris le problème avec au moins une phrase complète.")

        low = text.lower()
        if any(k in low for k in ["csv", "fichier de données", "données manquantes", "doublon", "qualité des données"]):
            domain = "qualite_donnees"
            objective = "Contrôler la qualité d'un fichier tabulaire et produire un rapport vérifiable."
        elif any(k in low for k in ["code", "programmation", "bug", "erreur python", "logiciel"]):
            domain = "developpement_logiciel"
            objective = "Analyser un problème logiciel et préparer une solution testable."
        else:
            domain = "logiciel_general"
            objective = "Décomposer le problème en exigences et proposer une solution logicielle testable."

        constraints = []
        if any(k in low for k in ["gratuit", "sans budget", "sans argent", "colab"]):
            constraints.append("Privilégier les outils gratuits et un environnement de prototypage accessible.")
        if any(k in low for k in ["confidentiel", "privé", "sensible", "secret"]):
            constraints.append("Limiter l'exposition des données et éviter de transmettre des contenus sensibles à des services externes.")
        if any(k in low for k in ["rapide", "simple", "débutant"]):
            constraints.append("Garder une première version simple à utiliser et à tester.")
        if not constraints:
            constraints.append("Rendre les hypothèses et les limites explicites ; commencer par un prototype de périmètre réduit.")

        if domain == "qualite_donnees":
            users = "Personnes ou équipes qui doivent contrôler des fichiers CSV."
            criteria = ["Détecter les cellules vides", "Repérer les lignes dupliquées", "Produire un rapport clair", "Gérer les fichiers invalides sans planter"]
        elif domain == "developpement_logiciel":
            users = "Développeurs ou personnes qui construisent des logiciels."
            criteria = ["Décomposer le problème", "Proposer une solution testable", "Rendre les hypothèses explicites", "Inclure des tests"]
        else:
            users = "Utilisateurs ayant un problème pouvant être traité par un logiciel."
            criteria = ["Répondre au besoin principal", "Rester réalisable", "Permettre de tester les résultats", "Expliquer les limites"]

        return ProblemSpec(text, domain, objective, users, constraints, criteria)


class IdeaEngine:
    """Produit trois approches heuristiques distinctes, adaptées au domaine détecté."""

    def generate(self, spec: ProblemSpec) -> List[Idea]:
        if spec.domain == "qualite_donnees":
            return [
                Idea(
                    name="Contrôle CSV minimal",
                    approach="Lire un CSV, calculer des indicateurs de qualité et écrire un rapport JSON.",
                    advantages=["Simple à exécuter", "Résultats faciles à vérifier", "Aucune API externe"],
                    limitations=["Prend en charge les fichiers CSV seulement", "Ne corrige pas automatiquement les données"],
                    complexity=1, template_key="csv_quality"
                ),
                Idea(
                    name="Contrôle CSV avec règles métier",
                    approach="Ajouter des règles configurables, par exemple formats d'e-mail ou champs obligatoires.",
                    advantages=["Détection adaptée au contexte", "Règles explicites et auditables"],
                    limitations=["Il faut définir les règles adaptées à chaque fichier", "Plus de configuration"],
                    complexity=3, template_key="csv_quality_rules"
                ),
                Idea(
                    name="Pipeline de validation reproductible",
                    approach="Séparer l'import, les contrôles, le rapport et les tests pour pouvoir étendre le système.",
                    advantages=["Architecture plus facile à étendre", "Tests indépendants par étape"],
                    limitations=["Plus de code pour une première version", "Demande une organisation plus stricte"],
                    complexity=4, template_key="csv_pipeline"
                ),
            ]

        return [
            Idea(
                name="Prototype minimal",
                approach="Construire une version très limitée qui résout le besoin principal et produit un résultat mesurable.",
                advantages=["Rapide à développer", "Facile à tester et à modifier"],
                limitations=["Peu de fonctionnalités", "Ne couvre pas tous les cas d'utilisation"],
                complexity=1, template_key="generic_mvp"
            ),
            Idea(
                name="Solution guidée par un workflow",
                approach="Découper le travail en étapes explicites avec des entrées, sorties et validations entre chaque étape.",
                advantages=["Parcours compréhensible", "Étapes plus faciles à diagnostiquer"],
                limitations=["Plus de logique à maintenir", "Peut être rigide si les besoins changent"],
                complexity=3, template_key="workflow"
            ),
            Idea(
                name="Solution extensible et testée",
                approach="Séparer le cœur métier, les interfaces et les tests pour ajouter des capacités progressivement.",
                advantages=["Plus facile à faire évoluer", "Meilleure séparation des responsabilités"],
                limitations=["Conception initiale plus longue", "Peut être excessive pour un petit besoin"],
                complexity=4, template_key="extensible"
            ),
        ]


class SolutionEvaluator:
    """Score heuristique explicable : les notes ne sont pas des mesures scientifiques."""

    WEIGHTS = {"usefulness": 0.30, "feasibility": 0.30, "testability": 0.25, "simplicity": 0.15}

    def evaluate(self, spec: ProblemSpec, ideas: List[Idea]) -> List[Evaluation]:
        results = []
        for idea in ideas:
            usefulness = 4 if spec.domain == "qualite_donnees" and idea.template_key == "csv_quality" else 3
            feasibility = max(1, 5 - idea.complexity // 2)
            testability = 5 if idea.template_key in {"csv_quality", "csv_quality_rules", "csv_pipeline", "generic_mvp"} else 4
            simplicity = max(1, 6 - idea.complexity)
            if spec.domain == "qualite_donnees" and idea.template_key.startswith("csv"):
                usefulness = min(5, usefulness + 1)
            values = {"usefulness": usefulness, "feasibility": feasibility, "testability": testability, "simplicity": simplicity}
            score = sum(values[k] * self.WEIGHTS[k] for k in self.WEIGHTS)
            reasons = [
                f"Utilité estimée : {usefulness}/5 selon la correspondance avec le besoin détecté.",
                f"Faisabilité estimée : {feasibility}/5 ; complexité annoncée {idea.complexity}/5.",
                f"Testabilité estimée : {testability}/5.",
                f"Simplicité estimée : {simplicity}/5."
            ]
            results.append(Evaluation(idea.name, usefulness, feasibility, testability, simplicity, round(score, 2), reasons))
        return sorted(results, key=lambda item: item.total_score, reverse=True)


class ArchitectureDesigner:
    def design(self, spec: ProblemSpec, idea: Idea) -> ArchitecturePlan:
        if idea.template_key.startswith("csv"):
            components = [
                {"name": "CSV Reader", "role": "Lire le fichier et vérifier la présence d'en-têtes."},
                {"name": "Quality Checks", "role": "Calculer les cellules manquantes et les lignes dupliquées."},
                {"name": "Report Builder", "role": "Construire un rapport JSON contenant les indicateurs."},
                {"name": "CLI", "role": "Permettre de lancer l'analyse sur un fichier choisi."},
            ]
            flow = ["Chemin du CSV", "Lecture et contrôle de structure", "Calcul des indicateurs", "Écriture du rapport JSON"]
            tests = ["CSV valide avec une ligne", "Cellules vides", "Lignes dupliquées", "Fichier vide ou sans en-têtes"]
            risks = ["Un CSV peut utiliser un séparateur ou un encodage inattendu.", "Une ligne identique est comptée comme doublon après la première occurrence."]
        else:
            components = [
                {"name": "Input Validator", "role": "Valider et normaliser les entrées."},
                {"name": "Core Logic", "role": "Implémenter le traitement principal décrit par la spécification."},
                {"name": "Result Formatter", "role": "Présenter les résultats et les limites."},
                {"name": "Test Suite", "role": "Vérifier les cas normaux et les cas limites."},
            ]
            flow = ["Entrée utilisateur", "Validation", "Traitement principal", "Résultat vérifiable"]
            tests = ["Entrée normale", "Entrée vide", "Entrée invalide", "Cas limite"]
            risks = ["Les exigences doivent être précisées avant de coder.", "Le code généré doit être vérifié par des tests réels."]
        return ArchitecturePlan(f"Architecture proposée — {idea.name}", components, flow, risks, tests)


class PrototypeBuilder:
    """V0.1 : génère un prototype exécutable pour le cas d'usage CSV seulement."""

    CSV_QUALITY_CODE = r'''#!/usr/bin/env python3
"""Prototype généré par INVENTRA AI : analyse de qualité d'un fichier CSV."""
import argparse
import csv
import json
from pathlib import Path


def analyze_csv(csv_path):
    """Analyse les cellules vides et les lignes dupliquées d'un CSV."""
    path = Path(csv_path)
    if not path.is_file():
        raise FileNotFoundError(f"Fichier introuvable : {path}")

    with path.open("r", encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        if not reader.fieldnames:
            raise ValueError("Le fichier CSV est vide ou ne contient pas d'en-têtes.")
        columns = list(reader.fieldnames)
        rows = list(reader)

    missing_by_column = {column: 0 for column in columns}
    duplicate_rows = 0
    seen = set()
    missing_cells = 0

    for row in rows:
        normalized = tuple((row.get(column) or "").strip() for column in columns)
        for column in columns:
            if not (row.get(column) or "").strip():
                missing_by_column[column] += 1
                missing_cells += 1
        if normalized in seen:
            duplicate_rows += 1
        else:
            seen.add(normalized)

    return {
        "file_name": path.name,
        "row_count": len(rows),
        "column_count": len(columns),
        "columns": columns,
        "missing_cells_total": missing_cells,
        "missing_by_column": missing_by_column,
        "duplicate_rows_after_first_occurrence": duplicate_rows,
        "quality_notes": [
            "Les cellules vides incluent les valeurs absentes ou composées d'espaces.",
            "Les doublons sont comptés après la première occurrence, après suppression des espaces aux extrémités."
        ]
    }


def main():
    parser = argparse.ArgumentParser(description="Analyse simple de la qualité d'un CSV")
    parser.add_argument("csv_file", help="Chemin vers le fichier CSV à analyser")
    parser.add_argument("--output", help="Chemin du rapport JSON (par défaut : <fichier>.quality_report.json)")
    args = parser.parse_args()

    report = analyze_csv(args.csv_file)
    output_path = Path(args.output) if args.output else Path(args.csv_file).with_suffix(".quality_report.json")
    output_path.write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
    print(json.dumps(report, ensure_ascii=False, indent=2))
    print(f"\nRapport enregistré : {output_path}")


if __name__ == "__main__":
    main()
'''

    def build(self, spec: ProblemSpec, idea: Idea, output_dir: str = ".") -> Dict[str, str]:
        if idea.template_key not in {"csv_quality", "csv_quality_rules", "csv_pipeline"}:
            return {
                "status": "not_supported_yet",
                "message": "La génération de code est activée uniquement pour le prototype d'analyse de qualité CSV dans cette V0.1. Les autres idées restent au stade de spécification.",
            }
        out = Path(output_dir)
        out.mkdir(parents=True, exist_ok=True)
        target = out / "inventra_csv_quality_prototype.py"
        ast.parse(self.CSV_QUALITY_CODE)
        target.write_text(self.CSV_QUALITY_CODE, encoding="utf-8")
        return {"status": "created", "file": str(target), "message": "Prototype CSV généré ; il doit encore être testé avec tes propres fichiers."}


class InventraAI:
    def __init__(self):
        self.analyzer = ProblemAnalyzer()
        self.idea_engine = IdeaEngine()
        self.evaluator = SolutionEvaluator()
        self.architect = ArchitectureDesigner()
        self.builder = PrototypeBuilder()

    def explore(self, description: str) -> Dict[str, Any]:
        spec = self.analyzer.analyze(description)
        ideas = self.idea_engine.generate(spec)
        evaluations = self.evaluator.evaluate(spec, ideas)
        by_name = {idea.name: idea for idea in ideas}
        selected_idea = by_name[evaluations[0].idea_name]
        architecture = self.architect.design(spec, selected_idea)
        return {"spec": spec, "ideas": ideas, "evaluations": evaluations, "selected_idea": selected_idea, "architecture": architecture}


def print_exploration(result: Dict[str, Any]) -> None:
    spec = result["spec"]
    print("=" * 68)
    print("INVENTRA AI — ANALYSE DU PROBLÈME")
    print("=" * 68)
    print(f"Domaine détecté : {spec.domain}")
    print(f"Objectif : {spec.objective}")
    print(f"Utilisateurs : {spec.users}")
    print("Contraintes :")
    for item in spec.constraints:
        print(f"  • {item}")
    print("\nSOLUTIONS PROPOSÉES")
    for idea in result["ideas"]:
        print(f"\n- {idea.name} (complexité estimée : {idea.complexity}/5)")
        print(f"  Approche : {idea.approach}")
        print(f"  Avantages : {', '.join(idea.advantages)}")
        print(f"  Limites : {', '.join(idea.limitations)}")
    print("\nCOMPARAISON HEURISTIQUE (pas une vérité objective)")
    for ev in result["evaluations"]:
        print(f"- {ev.idea_name}: score {ev.total_score}/5")
        for reason in ev.reasons:
            print(f"    · {reason}")
    print(f"\nApproche retenue par la grille : {result['selected_idea'].name}")
    print("\nARCHITECTURE PROPOSÉE")
    for component in result["architecture"].components:
        print(f"- {component['name']} : {component['role']}")
    print("Flux : " + " → ".join(result["architecture"].data_flow))
    print("Tests prévus : " + "; ".join(result["architecture"].tests))
    print("Risques : " + "; ".join(result["architecture"].risks))


print("INVENTRA AI V0.1 chargée.")
print("Prototype heuristique : aucun modèle génératif n'est connecté dans cette version.")
```

```python
# ============================================================
# Démonstration + tests automatiques
# ============================================================

import ast
from pathlib import Path
import tempfile

engine = InventraAI()
problem = (
    "Je veux créer un logiciel qui analyse un fichier CSV, détecte les cellules "
    "vides et les lignes dupliquées, puis produit un rapport clair. Il doit être gratuit."
)
result = engine.explore(problem)
print_exploration(result)


def run_self_tests():
    checks = []
    spec = ProblemAnalyzer().analyze(problem)
    checks.append(("analyse du problème", spec.domain == "qualite_donnees"))
    ideas = IdeaEngine().generate(spec)
    checks.append(("trois idées proposées", len(ideas) == 3))
    checks.append(("idées distinctes", len({idea.name for idea in ideas}) == 3))
    evaluations = SolutionEvaluator().evaluate(spec, ideas)
    checks.append(("évaluation de toutes les idées", len(evaluations) == 3))
    checks.append(("scores dans la plage 1–5", all(1 <= item.total_score <= 5 for item in evaluations)))
    architecture = ArchitectureDesigner().design(spec, ideas[0])
    checks.append(("architecture avec composants", len(architecture.components) >= 3))
    checks.append(("architecture avec tests", len(architecture.tests) >= 3))

    generated = PrototypeBuilder.CSV_QUALITY_CODE
    ast.parse(generated)
    checks.append(("syntaxe du prototype généré", True))
    namespace = {"__name__": "inventra_generated_test"}
    exec(compile(generated, "inventra_csv_quality_prototype.py", "exec"), namespace)
    with tempfile.TemporaryDirectory() as tmp:
        csv_file = Path(tmp) / "sample.csv"
        csv_file.write_text("nom,ville\nAwa,Dakar\nAwa,Dakar\nMoussa,\n", encoding="utf-8")
        report = namespace["analyze_csv"](csv_file)
        checks.append(("nombre de lignes", report["row_count"] == 3))
        checks.append(("détection d'une ligne dupliquée", report["duplicate_rows_after_first_occurrence"] == 1))
        checks.append(("détection d'une cellule vide", report["missing_cells_total"] == 1))
        checks.append(("nom de colonne dans le rapport", "ville" in report["missing_by_column"]))
        empty_file = Path(tmp) / "empty.csv"
        empty_file.write_text("", encoding="utf-8")
        try:
            namespace["analyze_csv"](empty_file)
            empty_csv_rejected = False
        except ValueError:
            empty_csv_rejected = True
        checks.append(("rejet d'un fichier sans en-têtes", empty_csv_rejected))

    print("\n" + "=" * 52)
    print("RÉSULTATS DES TESTS INVENTRA AI")
    print("=" * 52)
    for name, ok in checks:
        print(f"{'✅ RÉUSSI' if ok else '❌ ÉCHEC'} — {name}")
    passed = sum(ok for _, ok in checks)
    print(f"\nBilan : {passed}/{len(checks)} tests réussis.")
    if passed != len(checks):
        raise AssertionError("Au moins un test a échoué : examine les lignes signalées.")
    return checks


checks = run_self_tests()

output_dir = "/content/inventra_ai_output" if Path("/content").exists() else "inventra_ai_output"
build_result = PrototypeBuilder().build(result["spec"], result["selected_idea"], output_dir)
print("\nGÉNÉRATION DU PROTOTYPE")
print(build_result)
if build_result.get("status") == "created":
    print("\nFichier créé :", build_result["file"])
    print("Pour l'utiliser, lance ensuite une cellule avec :")
    print("!python /content/inventra_ai_output/inventra_csv_quality_prototype.py /chemin/vers/ton_fichier.csv")
```
